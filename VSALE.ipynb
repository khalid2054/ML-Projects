```python

df2.head(10)
```

```python
mode_year = df2['Year'].mode()[0]
df2['Year'] = df2['Year'].fillna(mode_year).astype(int)
```

```python
df2.dropna(subset=['Publisher'],inplace=True)
```

```python
final_df =df2.drop(columns=['Rank','Name','Other_Sales',])
```

```python
final_df = pd.get_dummies(columns=['Genre','Publisher','Platform'],drop_first=True,data= final_df)
```

```python
final_df.isna().sum()
```

```python
final_df = final_df.astype(int)
```

```python
final_df.columns
```

```python
final_df
```

## split

```python
X =final_df.drop('Global_Sales',axis = 1)
y = final_df['Global_Sales']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

## scaling 

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled_X_train = scaler.fit_transform(X_train)
scaled_X_test = scaler.transform(X_test)
```

##  model 1

```python
from sklearn.linear_model import LinearRegression
```

```python
model  = LinearRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error,mean_squared_error
```

```python
print(f'MSE of unsacled data is {mean_squared_error(y_pred,y_test)}')
```

```python
print(f'MAE of unsacled data is {mean_absolute_error(y_pred,y_test)}')
```

```python
print(f'R2 score of unsacled data is {r2_score(y_pred,y_test)}')
```

## model 2 

```python
model2  = LinearRegression()
model2.fit(scaled_X_train,y_train)
scaled_y_pred = model2.predict(scaled_X_test)
```

```python
print(f'MSE of sacled data is {mean_squared_error(scaled_y_pred,y_test)}')
```

```python
print(f'MAE of sacled data is {mean_absolute_error(scaled_y_pred,y_test)}')
```

```python
print(f'R2 score of sacled data is {r2_score(y_test,scaled_y_pred)}')
```
